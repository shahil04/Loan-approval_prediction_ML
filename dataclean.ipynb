```python
df['Gender'].isnull().sum()
df['Gender'].mode()[0]
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
# Replace the null values
for i in df.columns:
    # print(i)
    if df[i].dtypes =="O":
        # print(i)
        df[i] = df[i].fillna(df[i].mode()[0])
    else:
        df[i]=df[i].fillna(df[i].mean())
```

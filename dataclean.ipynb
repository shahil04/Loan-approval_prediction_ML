```python
# Replace the null values
for i in df.columns:
    # print(i)
    if df[i].dtypes =="O":
        # print(i)
        df[i] = df[i].fillna(df[i].mode()[0])
        df[i]=df[i].str.lower()
    else:
        df[i]=df[i].fillna(df[i].mean())
```

```python
# 2. data Preprocessing
df = df.drop("Loan_ID",axis=1)
```

```python
# N=0
# Y=1
# df["Loan_Status"] = df["Loan_Status"].map({"N":0,"Y":1})
from sklearn.preprocessing import LabelEncoder
l =LabelEncoder()
df["Loan_Status"] = l.fit_transform(df["Loan_Status"])
df.head()
```

## start the machine learning Training

```python
# split data into x and y
# split data in training and testing
# apply column transform for preprocessing in one line 
# Load the Classification model and fit the model
# Predict the test data
# accuracy check
# then start predictions
```

```python
df.head(1)
```

```python
df['Education'].unique()
```

```python
# split data into x and y
x= df.drop('Loan_Status',axis=1)
y =df['Loan_Status']

# train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

X_train.shape

# apply column transform for preprocessing in one line 
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler,OrdinalEncoder

columns_trans = ColumnTransformer([
        ('gender', OrdinalEncoder(), ['Gender']),
        ('married', OrdinalEncoder(), ['Married']),
        ('dependents', OrdinalEncoder(), ['Dependents']),
        ('education', OrdinalEncoder(), ['Education']),
        ('self_employed', OrdinalEncoder(), ['Self_Employed']),
        ('property_area',OneHotEncoder(sparse_output=False, handle_unknown='ignore'),['Property_Area']),
        ('scaler',StandardScaler(),['ApplicantIncome', 'CoapplicantIncome'])
    ],remainder='passthrough')
```

```python
# model load
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
from sklearn.pipeline import make_pipeline
final = make_pipeline(columns_trans,model)
final
```

```python
final.fit(X_train,y_train)
```

```python
print(final.score(X_train,y_train))
print(final.score(X_test,y_test))
```

```python
y_pred = final.predict(X_test)
```

```python
# Performance Matrix
# Measuring Performance metrics-Lost and Cost Function (MAE,MSE,RMSE,R2 Score)
# cost functions --> calculate erros
from sklearn.metrics import mean_absolute_error, mean_squared_error,root_mean_squared_error

print("MAE:",mean_absolute_error(y_test,y_pred))
print("MSE:",mean_squared_error(y_test,y_pred))
print("RMSE:",root_mean_squared_error(y_test,y_pred))
```

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
